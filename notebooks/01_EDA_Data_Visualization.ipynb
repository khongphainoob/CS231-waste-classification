# CS231 - Đồ án Thị Giác Máy Tính: 01. Exploratory Data Analysis (EDA)
## Phân loại rác thải rắn sinh hoạt đơn lẻ tại nguồn


```python
import os
import glob
import cv2
import numpy as np
import matplotlib.pyplot as plt
import yaml

with open('../configs/config.yaml', 'r', encoding='utf-8') as f:
    config = yaml.safe_load(f)

classes = list(config['dataset']['classes'].values())
print('Target Classes:', classes)
```

### 1. Thống kê số lượng ảnh theo từng lớp
Kiểm tra tính cân bằng dữ liệu giữa các lớp trên tập Train.


```python
train_dir = '../data/standard_split/train'
counts = {}
for c in classes:
    files = glob.glob(os.path.join(train_dir, c, '*.*'))
    counts[c] = len(files)

plt.figure(figsize=(9, 4))
colors = ['#D2691E', '#4682B4', '#708090', '#F5DEB3', '#00CED1', '#2E8B57']
plt.bar(counts.keys(), counts.values(), color=colors)
plt.title('Phân bố số lượng ảnh tập Train (560 ảnh / lớp chuẩn)')
plt.ylabel('Số lượng')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### 2. Trực quan hóa mẫu ảnh tiêu biểu của 6 lớp rác


```python
fig, axes = plt.subplots(2, 3, figsize=(12, 8))
axes = axes.flatten()

for i, c in enumerate(classes):
    img_files = glob.glob(os.path.join(train_dir, c, '*.*'))
    if img_files:
        bgr = cv2.imread(img_files[0])
        rgb = cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB)
        axes[i].imshow(rgb)
        bin_guide = config['dataset']['bin_mapping'].get(c, '')
        axes[i].set_title(f"Class: {c}\n[{bin_guide}]", fontsize=10)
    axes[i].axis('off')

plt.tight_layout()
plt.show()
```

### 3. Phân tích không gian màu: RGB vs HSV
Chứng minh tại sao HSV Histogram hiệu quả trong việc phân biệt rác hữu cơ (biological) và bìa carton (cardboard).


```python
bio_files = glob.glob(os.path.join(train_dir, 'biological', '*.*'))
if bio_files:
    bgr = cv2.imread(bio_files[0])
    rgb = cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB)
    hsv = cv2.cvtColor(rgb, cv2.COLOR_RGB2HSV)

    fig, ax = plt.subplots(1, 4, figsize=(16, 4))
    ax[0].imshow(rgb); ax[0].set_title('RGB Image'); ax[0].axis('off')
    ax[1].imshow(hsv[:, :, 0], cmap='hsv'); ax[1].set_title('Hue Channel (Sắc độ màu)'); ax[1].axis('off')
    ax[2].imshow(hsv[:, :, 1], cmap='viridis'); ax[2].set_title('Saturation Channel (Độ bão hòa)'); ax[2].axis('off')
    ax[3].imshow(hsv[:, :, 2], cmap='gray'); ax[3].set_title('Value Channel (Cường độ sáng)'); ax[3].axis('off')
    plt.tight_layout()
    plt.show()
```
